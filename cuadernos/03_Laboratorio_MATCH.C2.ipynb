```python
# @title Ejecuta primero: configuración y paleta de la clase
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams

FONDO='#101526'; TARJETA='#161C33'
MORADO='#7C8CFF'; ROSA='#FF4D8D'; CIAN='#22D3EE'
AMBAR='#FFC24B'; VERDE='#4ADE80'
GRIS='#9AA4B8'; TENUE='#46507A'; LINEA='#333B57'

def tema():
    rcParams['figure.facecolor']=FONDO
    rcParams['axes.facecolor']=TARJETA
    rcParams['savefig.facecolor']=FONDO
    rcParams['text.color']='white'
    rcParams['axes.labelcolor']=GRIS
    rcParams['xtick.color']=GRIS
    rcParams['ytick.color']=GRIS
    rcParams['axes.edgecolor']=LINEA
    rcParams['axes.titlecolor']='white'
    rcParams['font.size']=11
    rcParams['axes.spines.top']=False
    rcParams['axes.spines.right']=False

tema()
print('Listo.')
```

# TIP 1 · Propósito
## ¿Qué está haciendo esta gráfica aquí?

El mismo dato, tres propósitos. Fíjate que **no cambian los números**: cambia qué se destaca, qué se dice y qué se calla.

```python
meses = ['ene','feb','mar','abr','may','jun','jul','ago']
ventas = np.array([120, 118, 131, 127, 96, 134, 141, 138])

fig, ejes = plt.subplots(1, 3, figsize=(15, 4))

# EXPLICAR: contexto completo, sin énfasis
ejes[0].plot(meses, ventas, color=MORADO, lw=2, marker='o')
ejes[0].set_title('EXPLICAR\nVentas mensuales del semestre', loc='left', fontsize=12)

# PERSUADIR: un solo punto, el que sostiene la decision
col = [TENUE]*8; col[4] = ROSA
ejes[1].bar(meses, ventas, color=col)
ejes[1].annotate('paro de transportadores', xy=(4, 96), xytext=(1.2, 40),
                 color=ROSA, fontsize=10,
                 arrowprops=dict(arrowstyle='->', color=ROSA))
ejes[1].set_title('PERSUADIR\nMayo explica toda la caida del semestre', loc='left', fontsize=12)

# MONITOREAR: contra una meta, para saber si hay que actuar
meta = 125
sobre = ventas >= meta
ejes[2].bar(meses, ventas, color=[VERDE if s else ROSA for s in sobre])
ejes[2].axhline(meta, color='white', lw=1.2, ls='--')
ejes[2].text(7.4, meta+3, 'meta', color='white', ha='right', fontsize=10)
ejes[2].set_title('MONITOREAR\n3 de 8 meses por debajo de la meta', loc='left', fontsize=12)

plt.tight_layout(); plt.show()
print('Mismos ocho numeros. Tres graficas que no se pueden intercambiar.')
```

**Regla:** si no puedes decir el propósito en una frase, el gráfico todavía no existe.

**Y la audiencia:** la misma serie es *explicativa* para el equipo comercial y *persuasiva* para la junta que decide el presupuesto.

# TIP 2 · Tipo de gráfica
## ¿Qué forma sirve a ese propósito?

Cambia la pregunta en el selector y mira cómo cambia la forma correcta.

```python
# @title Selector: la pregunta manda sobre la forma
rng = np.random.default_rng(11)
cat = ['A','B','C','D','E']
val = np.array([32, 27, 19, 14, 8])
serie = np.cumsum(rng.normal(2, 5, 24)) + 60
x1 = rng.normal(50, 12, 120)
y1 = 0.7*x1 + rng.normal(0, 8, 120)
dist = rng.normal(70, 14, 400)

def dibujar(pregunta):
    fig, eje = plt.subplots(figsize=(8, 4))
    if pregunta.startswith('Comparar'):
        eje.bar(cat, val, color=TENUE); eje.bar(['A'], [32], color=MORADO)
        eje.set_title('Barras: la posicion comun hace la comparacion facil', loc='left')
    elif pregunta.startswith('Seguir'):
        eje.plot(serie, color=MORADO, lw=2)
        eje.set_title('Linea: la pendiente es el mensaje', loc='left')
    elif pregunta.startswith('Relacionar'):
        eje.scatter(x1, y1, s=28, color=MORADO, alpha=0.65, edgecolor='none')
        eje.set_title('Dispersion: dos variables, una relacion', loc='left')
    elif pregunta.startswith('Mostrar composicion'):
        eje.barh(['Total'], [32], color=MORADO)
        izq = 32
        for v, c in zip(val[1:], [CIAN, AMBAR, VERDE, ROSA]):
            eje.barh(['Total'], [v], left=izq, color=c); izq += v
        eje.set_title('Apilado: partes de un todo, y solo si son pocas', loc='left')
    else:
        eje.hist(dist, bins=18, color=MORADO)
        eje.set_title('Histograma: la forma de la distribucion', loc='left')
    plt.tight_layout(); plt.show()

OPCIONES = ['Comparar cantidades', 'Seguir una evolucion', 'Relacionar dos variables',
            'Mostrar composicion', 'Ver la distribucion']
try:
    import ipywidgets as W
    from IPython.display import display
    display(W.interactive(dibujar, pregunta=W.ToggleButtons(options=OPCIONES)))
except Exception:
    for o in OPCIONES:
        dibujar(o)
```

**Y la audiencia:** un boxplot es preciso, y para una audiencia no técnica es un gráfico que no existe. Si hay que explicar cómo se lee, ya perdiste los cinco segundos.

### 3.2 · El ojo agrupa por color antes que por forma

En el panel izquierdo el bloque aparece solo. En el derecho tienes que buscarlo.

```python
def rejilla(eje, patron, semilla=7):
    r = np.random.default_rng(semilla)
    cols, filas = 11, 10
    for f in range(filas):
        for c in range(cols):
            if patron == 'color':
                acento = c >= cols - 3
                cuadro = r.random() < 0.5
            else:
                cuadro = f >= filas - 4
                acento = r.random() < 0.42
            eje.scatter(c, -f, s=150,
                        marker='s' if cuadro else 'o',
                        color=ROSA if acento else TENUE)
    eje.set_xticks([]); eje.set_yticks([])
    for s in eje.spines.values():
        s.set_visible(False)

fig, (a, b) = plt.subplots(1, 2, figsize=(13, 5.5))
rejilla(a, 'color'); a.set_title('COLOR sobre forma', loc='left', fontsize=13)
rejilla(b, 'forma'); b.set_title('FORMA sobre color', loc='left', fontsize=13, color=GRIS)
plt.tight_layout(); plt.show()
print('El color es preatentivo: se procesa antes de que decidas mirar. La forma no.')
print('Por eso el color se reserva para lo que la audiencia debe ver primero.')
```

### 3.4 · Y la audiencia que no ve tu paleta

Ocho de cada cien hombres tienen alguna deficiencia en la visión del rojo y el verde.

```python
def simular_deuteranopia(hexcolor):
    """Aproximacion de como se percibe un color con deuteranopia."""
    h = hexcolor.lstrip('#')
    r, g, b = [int(h[i:i+2], 16)/255 for i in (0, 2, 4)]
    R = 0.625*r + 0.375*g
    G = 0.700*r + 0.300*g
    B = 0.300*g + 0.700*b
    return '#{:02X}{:02X}{:02X}'.format(*[int(max(0,min(1,v))*255) for v in (R,G,B)])

loc = ['Bogota','Cali','Medellin']
inc = np.array([85, 25, 60])
mala = ['#FF3B30','#34C759','#FFCC00']
buena = ['#1D4ED8','#DBEAFE','#60A5FA']

fig, ejes = plt.subplots(2, 2, figsize=(12, 7))
for fila, (pal, nombre) in enumerate([(mala,'Rojo-verde'), (buena,'Secuencial azul')]):
    for col, (p, vista) in enumerate([(pal,'vision tipica'),
                                      ([simular_deuteranopia(c) for c in pal],'deuteranopia')]):
        e = ejes[fila][col]
        e.bar(loc, inc, color=p)
        e.set_ylim(0, 100)
        e.set_title(nombre + ' - ' + vista, loc='left', fontsize=11)
plt.tight_layout(); plt.show()
print('En la fila de arriba, el mensaje desaparece. En la de abajo sobrevive.')
```

```python
# @title Jugar
from IPython.display import display, HTML

COLOR_TIP = {'1': MORADO, '2': CIAN, '3': ROSA, '4': AMBAR, '5': VERDE, 'MATCH': VERDE}
marcador = {'ok': 0, 'n': 0}

def _panel(color, titulo, cuerpo):
    return HTML("<div style='background:#161C33;border-left:5px solid " + color +
                ";padding:14px 18px;margin:8px 0;border-radius:4px;font-family:Segoe UI,sans-serif'>" +
                "<div style='color:" + color + ";font-weight:700;font-size:22px'>" + titulo + "</div>" +
                "<div style='color:#C9D1E4;font-size:14px;margin-top:8px;line-height:1.5'>" + cuerpo + "</div></div>")

def _contexto(d):
    return HTML("<div style='display:flex;gap:14px;font-family:Segoe UI,sans-serif;margin:6px 0 12px 0'>" +
                "<div style='flex:1;background:#14203A;padding:12px 16px;border-radius:4px'>" +
                "<div style='color:#22D3EE;font-size:11px;font-weight:700'>EL MENSAJE</div>" +
                "<div style='color:#fff;font-size:14px;margin-top:4px'>" + d['msg'] + "</div></div>" +
                "<div style='flex:1;background:#2A1329;padding:12px 16px;border-radius:4px'>" +
                "<div style='color:#FF4D8D;font-size:11px;font-weight:700'>LA AUDIENCIA</div>" +
                "<div style='color:#fff;font-size:14px;margin-top:4px'>" + d['aud'] + "</div></div></div>")

def revelar(n, respuesta):
    """n = numero de ronda (1..6). respuesta = '1'..'5' o 'MATCH'."""
    d = JUEGO[n - 1]
    respuesta = str(respuesta).upper()
    acierto = (respuesta == d['r'])
    marcador['n'] += 1
    if acierto:
        marcador['ok'] += 1
    col = COLOR_TIP[d['r']]
    if d['r'] == 'MATCH':
        titulo = 'MATCH' + ('   (acertaste)' if acierto else '   (no fallaba ninguno)')
    else:
        titulo = 'NO MATCH  ·  TIP ' + d['r'] + '  ·  ' + d['tip']
        if not acierto:
            titulo += '   (habias dicho ' + respuesta + ')'
    cuerpo = ('<b>Por que:</b> ' + d['por'] + '<br><br>' +
              '<b>Que habria que hacer:</b> ' + d['arr'] + '<br><br>' +
              '<span style="color:#9AA4B8"><b>Efecto en esa audiencia:</b> ' + d['ef'] + '</span>')
    display(_panel(col, titulo, cuerpo))
    print('Marcador: ' + str(marcador['ok']) + ' de ' + str(marcador['n']))

try:
    import ipywidgets as W
    estado = {'i': 0}
    cab = W.HTML(); ctx = W.Output(); graf = W.Output(); res = W.Output()
    opciones = W.ToggleButtons(options=['1','2','3','4','5','MATCH'], value=None)
    seguir = W.Button(description='Siguiente ronda', button_style='info')

    def pintar():
        i = estado['i']
        if i >= len(JUEGO):
            cab.value = "<h3 style='color:#7C8CFF;font-family:Segoe UI'>Fin del juego</h3>"
            opciones.layout.display = 'none'; seguir.layout.display = 'none'
            for o in (ctx, graf):
                o.clear_output()
            with res:
                res.clear_output()
                display(_panel(MORADO, 'Marcador final: ' + str(marcador['ok']) + ' de ' + str(marcador['n']),
                               'Ninguna de las seis rondas se resolvia mirando los datos. '
                               'Todas se resolvian mirando la audiencia.'))
            return
        d = JUEGO[i]
        cab.value = ("<div style='font-family:Segoe UI,sans-serif;color:#FFC24B;"
                     "font-weight:700;font-size:13px'>RONDA " + str(i+1) + " DE 6</div>")
        opciones.value = None
        with ctx:
            ctx.clear_output(); display(_contexto(d))
        with graf:
            graf.clear_output(); d['f']()
        with res:
            res.clear_output()

    def contestar(c):
        if c['new'] is None:
            return
        with res:
            res.clear_output()
            revelar(estado['i'] + 1, c['new'])

    def avanzar(_):
        estado['i'] += 1; pintar()

    opciones.observe(contestar, names='value')
    seguir.on_click(avanzar)
    pintar()
    display(W.VBox([cab, ctx, graf, opciones, seguir, res]))
except Exception:
    print('Modo texto. Para cada ronda n: ejecuta JUEGO[n-1]["f"]() y luego revelar(n, tu_respuesta)')
    for k, d in enumerate(JUEGO, 1):
        print(str(k) + '. mensaje: ' + d['msg'] + ' | audiencia: ' + d['aud'])
```
