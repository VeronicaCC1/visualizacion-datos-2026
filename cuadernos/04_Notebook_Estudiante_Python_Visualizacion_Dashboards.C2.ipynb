## 6 · Altair / Vega-Lite

**Vocabulario:** `data` → `mark` → `encoding` (x, y, color, size, tooltip) →
`scale` → `transform` → `layer`/`facet` → `interaction`. La misma filosofía
de "grammar of graphics" que ggplot2.

### 6.1 · Un scatter declarativo

````python
chart = alt.Chart(df_2007).mark_circle(opacity=0.7).encode(
    x=alt.X("gdpPercap", scale=alt.Scale(type="log"), title="PIB per cápita (log)"),
    y=alt.Y("lifeExp", title="Expectativa de vida"),
    color=alt.Color("continent", title="Continente"),
    size=alt.Size("pop", title="Población", scale=alt.Scale(range=[20, 800])),
    tooltip=["country", "lifeExp", "gdpPercap", "pop"],
).properties(width=450, height=320, title="PIB per cápita vs. expectativa de vida — 2007")

chart
````

## 9 · Diseño de dashboards

Principios vistos en clase: jerarquía visual, proximidad/alineación,
consistencia, color intencional, densidad/carga cognitiva, títulos
informativos.

Un dashboard es **útil** si responde preguntas reales; **legible** si se
entiende en <10 segundos; **efectivo** si lleva a una decisión; **engañoso**
