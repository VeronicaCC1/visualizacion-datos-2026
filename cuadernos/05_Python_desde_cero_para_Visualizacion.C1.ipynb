### 🎓 1.1 · Una variable es una caja con nombre

Cuando escriben `duracion_min = 12`, están diciendo: "guarda el número 12 en una caja, y ponle de nombre `duracion_min`". De ahí en adelante, cada vez que escriban `duracion_min`, Python va y mira qué hay dentro de esa caja.

El signo `=` aquí **no significa "es igual a"** como en matemáticas. Significa "guarda esto en la caja de la izquierda". Por eso `duracion_min = 12` se lee "duracion_min recibe 12", no "duracion_min es igual a 12".

```python
# ▶️ Creamos tres variables sobre un solo viaje en bicicleta
duracion_min = 12
estacion = "Chapinero"
activo = True

print(duracion_min)
print(estacion)
print(activo)
```

🧐 Tres cajas, tres nombres, tres contenidos distintos. Fíjense en algo: `estacion` va entre comillas y `duracion_min` no. Eso no es casualidad — tiene que ver con el **tipo de dato** de cada una, que es el siguiente concepto.

💡 Los nombres de variable en Python no pueden tener espacios ni empezar con un número. `duracion_min` funciona; `duracion min` o `1viaje` no. La convención en Python es usar minúsculas y guion bajo (`_`) para separar palabras — así es como está escrito todo este cuaderno.

### 🎓 1.2 · Los cuatro tipos de dato básicos

Python no guarda "un valor": guarda un valor **de un tipo específico**, y ese tipo determina qué se puede hacer con él. Los cuatro que van a usar todo el semestre:

| Tipo | Nombre en Python | Ejemplo | Se reconoce por |
|---|---|---|---|
| Entero | `int` | `12` | Un número sin punto decimal |
| Decimal | `float` | `12.5` | Un número con punto decimal |
| Texto | `str` | `"Chapinero"` | Va entre comillas, simples o dobles |
| Verdadero/falso | `bool` | `True` | Solo existen `True` y `False`, con mayúscula inicial |

La función `type()` les dice, sin adivinar, de qué tipo es cualquier variable.

```python
print(type(duracion_min))
print(type(estacion))
print(type(activo))
```

🎯 Esto no es un detalle técnico que se olvida después de esta clase. El tipo de una columna es EXACTAMENTE lo que van a revisar con `df.dtypes` cada vez que carguen un archivo nuevo, todo el semestre. ### 🎓 1.3 · Operadores: hacer cuentas con las cajas

Los operadores aritméticos funcionan como en una calculadora, con una diferencia que vale la pena memorizar: `/` siempre da un resultado decimal, y `//` da la división "entera" (sin el resto).

| Operador | Qué hace | Ejemplo | Resultado |
|---|---|---|---|
| `+` `-` `*` | Suma, resta, multiplicación | `10 * 2` | `20` |
| `/` | División (con decimales) | `15 / 4` | `3.75` |
| `//` | División entera (sin decimales) | `15 // 4` | `3` |
| `%` | El residuo de la división | `15 % 4` | `3` |
| `**` | Potencia | `2 ** 3` | `8` |

```python
duracion_min = 135  # un viaje largo

horas = duracion_min // 60
minutos_restantes = duracion_min % 60

print(f"{duracion_min} minutos son {horas} horas y {minutos_restantes} minutos")
```

🧐 `//` les dio las horas completas, y `%` les dio lo que sobra después de esas horas — los dos operadores trabajan juntos para convertir minutos en "horas y minutos", algo que van a necesitar cada vez que una columna de tiempo venga en una sola unidad. (Ignoren por ahora el `f` antes de las comillas del `print` — llega en la Parte 2.)

### 🎓 1.4 · Operadores de comparación: preguntas de sí o no

Estos no calculan un número: responden `True` o `False`.

| Operador | Pregunta que hace |
|---|---|
| `==` | ¿Son iguales? |
| `!=` | ¿Son distintos? |
| `>` `<` | ¿Mayor que? ¿Menor que? |
| `>=` `<=` | ¿Mayor o igual? ¿Menor o igual? |

💡 `==` (dos signos) compara. `=` (un signo) asigna. Confundirlos es el error más común de quien empieza en Python — y Python no siempre avisa con un error claro cuando lo hacen mal.

```python
duracion_min = 45

print(duracion_min > 30)
print(duracion_min == 45)
print(duracion_min == 12)
```

### 🎓 2.1 · Cadenas de texto (`str`)

Un texto en Python se llama **cadena** (*string*), y va siempre entre comillas — simples `'...'` o dobles `"..."`, da igual cuál usen, mientras abran y cierren con la misma.

Para pegar dos textos, se usa `+`. Pero hay un detalle: `+` entre textos NO agrega espacios ni convierte números automáticamente.

```python
estacion = "Chapinero"
mensaje = "La estación de origen fue " + estacion + "."
print(mensaje)
```

### 🎓 2.2 · f-strings: la forma moderna de construir texto

Escribir mensajes a pura concatenación con `+` se vuelve incómodo apenas hay números de por medio (porque `+` no mezcla texto y números sin ayuda). La solución que van a usar todo el semestre son las **f-strings**: una cadena con una `f` justo antes de las comillas, donde cualquier cosa entre llaves `{ }` se reemplaza por su valor.

```python
estacion = "Chapinero"
duracion_min = 12

mensaje = f"El viaje que salió de {estacion} duró {duracion_min} minutos."
print(mensaje)
```

🧐 Dentro de `{ }` no hace falta convertir nada a texto a mano, ni preocuparse por espacios: la `f` se encarga. Esto es exactamente lo que van a escribir cada vez que le pongan título a un gráfico con un número calculado adentro — por ejemplo, `f"Duración promedio: {promedio:.1f} minutos"`.

💡 Ese `:.1f` que acaban de ver es un formato: le dice a Python "muéstrame este número con un solo decimal". Sin él, un promedio podría salir como `14.285714285714286` — técnicamente correcto, pero imposible de leer en un título.

```python
promedio_duracion = 14.285714285714286

print(f"Duración promedio: {promedio_duracion:.1f} minutos")
print(f"Duración promedio: {promedio_duracion:.0f} minutos")
```

### 🎓 3.1 · Listas: varios valores, un solo nombre

Hasta ahora, una variable guardaba un solo valor. Una **lista** guarda varios, en un orden fijo, entre corchetes `[ ]` y separados por comas.

```python
duraciones = [12, 8, 15, 22, 9, 30]

print(duraciones)
print(len(duraciones))
```

🧐 `len(...)` (de *length*, longitud) cuenta cuántos elementos tiene la lista — aquí, seis viajes.

Para acceder a un elemento puntual, se usa el corchete con su posición. **Ojo:** Python cuenta desde 0, no desde 1. El primer elemento está en la posición `0`.

```python
print(duraciones[0])   # el primer viaje
print(duraciones[2])   # el tercer viaje
print(duraciones[-1])  # el último viaje, sin saber cuántos hay
```

💡 `[-1]` es un atajo muy usado: "el último", sin tener que calcular `len(duraciones) - 1`. `[-2]` sería el penúltimo.

Para agregar un elemento nuevo al final, se usa `.append(...)`.

```python
duraciones.append(18)
print(duraciones)
```

### 🎓 3.2 · Diccionarios: cuando cada valor necesita un nombre

Una lista guarda valores en orden, pero no dice qué es cada uno. Si quieren describir **un solo viaje** con varios datos (estación, duración, tipo de usuario), lo natural no es una lista — es un **diccionario**: pares de `llave: valor`, entre llaves `{ }`.

```python
viaje = {
    "estacion": "Chapinero",
    "duracion_min": 12,
    "tipo_usuario": "mensual",
}

print(viaje["estacion"])
print(viaje["duracion_min"])
```

🧐 En una lista se accede por **posición** (`[0]`); en un diccionario se accede por **nombre** (`["estacion"]`). Esa diferencia importa: dentro de un rato van a ver que cada columna de una tabla de datos se comporta exactamente como la llave de un diccionario.

### 🎓 4.1 · `if`: que el código decida solo

Un bloque `if` ejecuta código **solo si** una comparación es `True`. Fíjense en dos detalles de forma que Python exige siempre: dos puntos `:` al final de la línea del `if`, y la línea siguiente **indentada** (con sangría, normalmente 4 espacios — Colab la pone automática al presionar Enter).

```python
duracion_min = 45

if duracion_min > 60:
    print("Viaje largo")
elif duracion_min > 20:
    print("Viaje medio")
else:
    print("Viaje corto")
```

🧐 Python evalúa las condiciones en orden, de arriba hacia abajo, y ejecuta solo la primera que resulte verdadera. Con `duracion_min = 45`, la primera condición (`> 60`) es falsa, la segunda (`> 20`) es verdadera — así que imprime "Viaje medio" y ni siquiera revisa el `else`.

💡 `elif` es la contracción de "else if" ("si no, entonces si"). Pueden encadenar tantos `elif` como necesiten; el `else` final es opcional, y atrapa "todo lo que no cumplió ninguna condición anterior".

### 🎓 4.2 · `for`: repetir una acción para cada elemento

Un bloque `for` recorre una lista, uno por uno, y repite el código de adentro para cada elemento.

```python
duraciones = [12, 8, 15, 22, 9, 30]

for duracion in duraciones:
    print(f"Este viaje duró {duracion} minutos")
```

🧐 En cada vuelta, la variable `duracion` toma el valor del siguiente elemento de la lista — primero `12`, luego `8`, y así hasta el final. El nombre `duracion` no es especial: pudieron haberlo llamado `x` o `valor`, pero un nombre que describe qué representa hace el código más fácil de leer (por ustedes mismos, dos semanas después).

### 🎓 4.3 · `if` + `for`: la combinación que van a usar todo el semestre

Juntos, sirven para responder preguntas como "¿cuántos de estos viajes duraron más de 15 minutos?" — algo que van a hacer una y otra vez con datos reales, antes de que `pandas` lo haga en una sola línea.

```python
duraciones = [12, 8, 15, 22, 9, 30]
contador_largos = 0

for duracion in duraciones:
    if duracion > 15:
        contador_largos = contador_largos + 1

print(f"{contador_largos} de {len(duraciones)} viajes duraron más de 15 minutos")
```

🧐 `contador_largos = 0` antes del `for` es un **acumulador**: una caja que arranca en cero y va sumando 1 cada vez que se cumple la condición, vuelta tras vuelta. Este patrón —acumulador antes del `for`, condición adentro— es la base de casi cualquier cálculo manual sobre una lista, y es exactamente lo que `pandas` va a reemplazar en la Parte 6.

### 🎓 5.1 · Por qué empaquetar código

Ya escribieron, en la Parte 4, código que clasifica una duración en corta/media/larga. Si tuvieran que clasificar 200 duraciones distintas, no van a copiar y pegar ese `if` 200 veces — van a **empaquetarlo** en una función, y llamarla 200 veces.

Una función se define con `def`, un nombre, paréntesis con los datos que necesita (los **parámetros**), dos puntos, y el código indentado debajo. `return` es lo que la función entrega de vuelta.

```python
def clasificar_duracion(minutos):
    if minutos > 60:
        return "Viaje largo"
    elif minutos > 20:
        return "Viaje medio"
    else:
        return "Viaje corto"


print(clasificar_duracion(45))
print(clasificar_duracion(75))
print(clasificar_duracion(10))
```

🧐 Definieron la función una sola vez, y la llamaron tres veces con datos distintos. Fíjense que es literalmente el mismo `if/elif/else` de la Parte 4, solo que ahora vive adentro de una función y puede reutilizarse con cualquier número, sin volver a escribirlo.

💡 `print(...)` y `return` no son lo mismo. `print` solo *muestra* algo en pantalla; `return` *entrega* un valor que se puede guardar en una variable y seguir usando. Una función que solo tiene `print` adentro no le sirve de nada al código que la llama, más allá de mostrar texto.

### 🎓 5.2 · Una función real, para el resto del cuaderno

Ahora que ya saben leer una función, aquí hay una que van a usar en el reto del final: compara lo que ustedes calcularon contra lo que se esperaba, y les dice si acertaron.

```python
def revisar(obtenido, esperado, pista=""):
    """Compara tu resultado con el esperado y te dice si acertaste."""
    if obtenido == esperado:
        print("✅ ¡Correcto!")
    else:
        print("❌ Todavía no.")
        print("   Tu resultado:", obtenido)
        print("   Se esperaba :", esperado)
        if pista:
            print("   💡 Pista:", pista)


# Pruébenla ustedes mismos:
revisar(2 + 2, 4)
revisar(2 + 2, 5, "Revisen la suma")
```

🧐 `revisar` tiene tres parámetros: `obtenido`, `esperado`, y `pista` — este último con un valor por defecto (`=""`), así que pueden llamarla sin dar ese tercer dato si no quieren dar pista. Adentro, es exactamente el mismo `if/else` de la Parte 4.1, solo que empaquetado. Guárdenla en la memoria: la van a usar sin volver a escribirla.

### 🎓 6.1 · Una tabla, hecha con lo que ya saben

Con listas y diccionarios ya pueden representar algo parecido a una tabla: una **lista de diccionarios**, donde cada diccionario es una fila.

```python
viajes = [
    {"estacion": "Chapinero", "duracion_min": 12, "tipo_usuario": "mensual"},
    {"estacion": "Usaquén", "duracion_min": 22, "tipo_usuario": "ocasional"},
    {"estacion": "Chapinero", "duracion_min": 8, "tipo_usuario": "ocasional"},
    {"estacion": "Suba", "duracion_min": 30, "tipo_usuario": "mensual"},
    {"estacion": "Usaquén", "duracion_min": 15, "tipo_usuario": "mensual"},
]

print(viajes[0])
print(viajes[0]["duracion_min"])
```

🧐 `viajes` es una lista con 5 elementos, y cada elemento es un diccionario — una fila con tres columnas: `estacion`, `duracion_min`, `tipo_usuario`. `viajes[0]` es la primera fila completa; `viajes[0]["duracion_min"]` es una celda puntual.

Ahora calculen el promedio de duración, con las herramientas que ya tienen: un acumulador y un `for`.

```python
suma = 0
for viaje in viajes:
    suma = suma + viaje["duracion_min"]

promedio = suma / len(viajes)
print(f"Duración promedio: {promedio:.1f} minutos")
```

Funciona perfecto para 5 viajes. Ahora imaginen hacerlo para 5.000 viajes, o para calcular el promedio de duración **por cada estación por separado**. El código no cambiaría de idea, pero se volvería largo y fácil de dañar por accidente.

### 🎓 6.3 · Cargar un dataset real

De aquí en adelante trabajamos con `viajes_marzo`: un mes completo de viajes en bicicleta, ya limpio, para que el resto del cuaderno se concentre en graficar y no en depurar. Lo generamos aquí mismo, para no depender de subir ningún archivo.

```python
# ▶️ EJECUTA ESTA CELDA — genera el dataset que usaremos en el resto del cuaderno

import numpy as np

rng = np.random.default_rng(7)
n = 220

estaciones = ["Chapinero", "Usaquén", "Teusaquillo", "Suba", "Kennedy"]
tipos_usuario = ["mensual", "ocasional"]

estacion = rng.choice(estaciones, size=n, p=[0.28, 0.22, 0.18, 0.17, 0.15])
tipo_usuario = rng.choice(tipos_usuario, size=n, p=[0.62, 0.38])
es_mensual = tipo_usuario == "mensual"

hora = np.where(
    es_mensual,
    np.clip(rng.normal(rng.choice([7, 18], size=n), 1.2, n).round(), 5, 22),
    rng.integers(8, 21, n),
).astype(int)

duracion_min = rng.normal(np.where(es_mensual, 13, 22), np.where(es_mensual, 4, 12), n).clip(2, None).round(1)
edad_usuario = rng.normal(np.where(es_mensual, 31, 26), np.where(es_mensual, 6, 9), n).clip(16, 70).round().astype(int)

viajes_marzo = pd.DataFrame({
    "estacion": estacion, "hora": hora, "duracion_min": duracion_min,
    "edad_usuario": edad_usuario, "tipo_usuario": tipo_usuario,
})

print(f"Listo: {viajes_marzo.shape[0]} viajes, {viajes_marzo.shape[1]} columnas.")
viajes_marzo.head()
```

🧐 No hace falta que entiendan cada línea de esta celda — está usando herramientas de `numpy` que no hemos visto, solo para fabricar datos realistas rápido. Lo que sí importa es el resultado: `viajes_marzo`, una tabla de 220 filas con cinco columnas, lista para explorar. `.head()` muestra las primeras 5 filas, para no imprimir las 220 de una vez.

### 🎓 9.1 · Una librería construida encima de otra

`seaborn` no reemplaza a `matplotlib`: está construido encima de él, y sabe leer directamente una tabla de `pandas`. La diferencia práctica: un boxplot **por categoría** —una caja para cada estación, por ejemplo— les tomaría varias líneas armar a mano con `matplotlib`. Con `seaborn`, es una sola.

```python
import seaborn as sns

sns.set_theme(style="whitegrid")  # un estilo visual, una sola vez, para todo el cuaderno

plt.figure(figsize=(9, 5))
sns.boxplot(data=viajes_marzo, x="estacion", y="duracion_min")
plt.title("Duración de los viajes, por estación")
plt.show()
```

🧐 `data=viajes_marzo` le dice a `seaborn` de qué tabla sacar los datos; `x=` y `y=` son simplemente los **nombres** de las columnas, como texto. `seaborn` se encarga de separar los datos por cada valor distinto de `estacion` y dibujar una caja por cada uno — sin que ustedes tengan que filtrar nada a mano.

### 🎓 9.2 · Dispersión: ¿dos columnas numéricas se mueven juntas?

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=viajes_marzo, x="hora", y="duracion_min", hue="tipo_usuario", alpha=0.6)
plt.title("Duración del viaje según la hora del día")
plt.show()
```

🧐 Cada punto es un viaje. `hue="tipo_usuario"` colorea los puntos según esa columna — así, sin filtrar nada, se ven al mismo tiempo el patrón general y si ese patrón cambia entre los dos tipos de usuario. `alpha=0.6` hace los puntos semitransparentes, para que donde se amontonan muchos se vea más oscuro.

💡 `hue` es el parámetro que más veces van a usar en `seaborn`. Cualquier gráfico —dispersión, barras, boxplot— acepta `hue=` para separar por una categoría más, sin escribir un `for` ni un filtro.

### 10.1 · Distribución — ¿cómo se reparten los valores?

```python
plt.figure(figsize=(8, 4))
sns.histplot(data=viajes_marzo, x="duracion_min", kde=True)
plt.title("Distribución de la duración de los viajes")
plt.show()
```

🧐 `kde=True` agrega una curva suave encima del histograma, que estima la forma de la distribución. Si la curva tiene una cola larga hacia la derecha, hay pocos viajes muy largos que estiran el promedio hacia arriba — algo que el promedio solo no dice.

### 10.2 · Atípicos — ¿hay valores extremos?

```python
plt.figure(figsize=(8, 3))
sns.boxplot(data=viajes_marzo, x="duracion_min")
plt.title("¿Hay viajes con duración atípica?")
plt.show()
```

🎯 Un valor atípico no es automáticamente un error. Un viaje inusualmente largo puede ser un error de registro — o puede ser alguien que usó la bicicleta todo el día. La regla que se lleven de este curso: un atípico se investiga antes de decidir qué hacer con él, nunca se borra por reflejo.

### 10.3 · Relaciones — ¿dos columnas numéricas se mueven juntas?

```python
correlaciones = viajes_marzo[["duracion_min", "edad_usuario", "hora"]].corr()

plt.figure(figsize=(5, 4))
sns.heatmap(correlaciones, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlación entre columnas numéricas")
plt.show()
```

🧐 `.corr()` calcula, para cada par de columnas numéricas, un número entre -1 y 1: cerca de 1 significa que suben juntas; cerca de -1, que una sube cuando la otra baja; cerca de 0, que no hay una relación lineal clara. `annot=True` escribe el número dentro de cada celda del mapa de calor, para no depender solo del color.

### 10.4 · Grupos — ¿el comportamiento cambia entre categorías?

```python
promedio_por_tipo = viajes_marzo.groupby("tipo_usuario")["duracion_min"].mean()
print(promedio_por_tipo)

plt.figure(figsize=(6, 4))
sns.barplot(x=promedio_por_tipo.index, y=promedio_por_tipo.values)
plt.title("Duración promedio, por tipo de usuario")
plt.ylabel("duración promedio (min)")
plt.show()
```

🧐 `.groupby("tipo_usuario")` separa la tabla en dos montones (mensual / ocasional); `["duracion_min"].mean()` calcula el promedio dentro de cada montón. Es el mismo par de ideas de la Parte 6 —seleccionar una columna, resumir con una función— aplicado después de agrupar.
